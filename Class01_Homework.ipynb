## 第 2 题｜检测文件编码

用 chardet 检测 `datasets/CRSP_Daily_Market_1926_2025.csv` 的编码，并打印检测到的编码名称和置信度。

```python
import chardet

# 直接复用课堂上定义过的 check_encoding() 函数
def check_encoding(filename):
    """检测文件编码，返回 (编码名称, 置信度)"""
    rawdata = open(filename, 'rb').read()   # 以二进制模式读取文件的原始字节
    result = chardet.detect(rawdata)        # chardet 根据字节内容推测编码
    return result['encoding'], result['confidence']

encoding, confidence = check_encoding('datasets/CRSP_Daily_Market_1926_2025.csv')
print(f"Encoding: {encoding}, Confidence: {confidence}")
```

## 第 5 题｜数据的基本信息与统计摘要

用 `info()` 和 `describe()` 查看 `crsp` 的列类型和统计摘要。
**思考：** vwretd、sprtrn 这些列名可能是什么意思？为什么 sprtrn、spindx 的非缺失值个数明显更少？

```python
crsp.info()  # 每一列的列名、非缺失值个数、数据类型
```

```python
crsp.describe()  # 数值列的统计摘要：个数、均值、标准差、最大/最小值等
```

**思考题回答：**

- `vwretd` / `vwretx`：市值加权（value-weighted）市场日收益率，含 / 不含股息；`ewretd` / `ewretx`：等权重（equal-weighted）日收益率；`sprtrn`：标普 500 日收益率；`spindx`：标普 500 指数点位。
- `sprtrn`、`spindx` 的非缺失值只有约 1.6 万条（15,981 / 15,982），明显少于全表的 26,301 条：因为标普 500 的相关数据**从 1962 年 7 月才开始有**（首条有效数据为 1962-07-03），之前年份为缺失值。

## 第 7 题｜选择行

① 用 `iloc` 选出第 0 行到第 4 行、第 0 列到第 2 列的数据；
② 用 `at` 取出第 1 行 `vwretd` 列的数值（第 0 行是 1925-12-31，当日还没有收益率，为缺失值）；
③ 用 `loc` 选出第 0 行到第 4 行、且只包含 `DlyCalDt` 和 `vwretd` 两列的数据。注意 iloc 和 loc 切片的区别。

```python
# ① iloc 按位置（行号、列号）选取：第 0~4 行、第 0~2 列
crsp.iloc[0:5, 0:3]
```

```python
# ② at 按"行索引 + 列名"取单个元素：第 1 行 vwretd 列
crsp.at[1, 'vwretd']
```

```python
# ③ loc 按标签切片：第 0~4 行、只取 DlyCalDt 和 vwretd 两列
# 注意：loc 的切片【包含】右端点（0:4 取到第 4 行），与 iloc 相反
crsp.loc[0:4, ['DlyCalDt', 'vwretd']]
```

## 第 8 题｜日期处理三步曲

① 把 `DlyCalDt` 列从文本转换为 datetime 格式（格式为 %Y-%m-%d）；
② 按 `DlyCalDt` 列升序排序；
③ 把 `DlyCalDt` 列设为索引。每一步都运行显示一下 `crsp`，确认结果正确。

```python
# 第①步：文本 → datetime（日期格式为 %Y-%m-%d，如 1925-12-31）
crsp['DlyCalDt'] = pd.to_datetime(crsp['DlyCalDt'], format='%Y-%m-%d')
crsp.head(3)  # 确认日期列已变成真正的日期格式
```

```python
# 第②步：按 DlyCalDt 升序排序（从旧到新）
crsp = crsp.sort_values(by='DlyCalDt', ascending=True)
crsp.head(3)  # 确认：最早的数据排在最上面
```

```python
# 第③步：把 DlyCalDt 设为索引——之后就可以按时间直接筛选
crsp.set_index('DlyCalDt', inplace=True)
crsp
```

## 第 9 题｜按时间筛选

用上题的 `crsp`，筛选出 2020 年全年的数据（2020 年爆发了新冠疫情，美股出现剧烈波动），把结果赋值给变量 `crsp_2020`。检查筛选结果的行数大约是多少个交易日？

```python
# DlyCalDt 已是索引，直接用年份 '2020' 切片筛选全年
crsp_2020 = crsp.loc['2020']
print('2020年交易日数：', crsp_2020.shape[0])
crsp_2020.head()  # 查看筛选结果的前几行
```

**检查结果：** 2020 年共筛选出 **253 个交易日**（2020-01-02 至 2020-12-31），与美股 2020 年实际交易日数量一致。

## 第 10 题｜保存数据

把筛选出的 `crsp_2020` 保存为 `datasets/CRSP_2020.csv`，要求：
① 把日期索引一并写入文件；② 用 Excel 直接打开时不出现乱码。保存后用 `pd.read_csv()` 重新读回来，验证文件是否正确生成。

```python
# 保存：index=True 把日期索引一并写入；encoding='utf-8-sig' 保证用 Excel 打开中文不乱码
crsp_2020.to_csv('datasets/CRSP_2020.csv', index=True, encoding='utf-8-sig')
```

```python
# 读回验证：文件能正常读回来，说明保存成功
pd.read_csv('datasets/CRSP_2020.csv').head()
```
